# Team 3 — Exploratory Data Analysis & Visualizations

This notebook demonstrates automated EDA, statistical distribution analysis, outlier detection, and correlation matrices using **AutoAnalyst AI**.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

from autoanalyst.data_loading.loader import load_dataset
from autoanalyst.eda.analyzer import (
    get_correlation_matrix,
    get_distribution_overview,
    get_outliers_summary,
)

dataset_path = (
    Path("../data/sample/example.csv")
    if Path("../data/sample/example.csv").exists()
    else Path("data/sample/example.csv")
)
df = load_dataset(dataset_path)
print(f"Loaded dataset: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

## 1. Summary Statistics & Shape of Distribution

```python
dist_df = get_distribution_overview(df)
dist_df
```

## 2. Correlation Analysis

```python
corr = get_correlation_matrix(df, method="pearson")
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="Blues", fmt=".2f")
plt.title("Pearson Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 3. Outliers Summary

```python
outlier_summary = get_outliers_summary(df, method="iqr")
outlier_summary
```
